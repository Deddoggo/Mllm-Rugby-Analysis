# Spatial Augmentation

```python
import torch
from torchvision.transforms import v2

spatial_augmentation = v2.Compose([
    v2.RandomHorizontalFlip(p=0.5),

    v2.RandomApply([
        v2.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05)
    ], p=0.8),

    v2.RandomErasing(p=0.3, scale=(0.02, 0.15), ratio=(0.3, 3.3), value='random')
])
```

# Temporal Augmentation


```python
import random
import numpy as np

def temporal_subsample(total_frames, num_frames_needed=16):
    """
    Randomly subsample frames from a video clip.
    """
    if total_frames <= num_frames_needed:
        return np.linspace(0, total_frames-1, num_frames_needed, dtype=int)
    
    max_stride = total_frames // num_frames_needed
    stride = random.randint(1, max_stride)

    max_start_idx = total_frames - (num_frames_needed * stride)
    start_idx = random.randint(0, max_start_idx)

    frame_indices = [start_idx + i * stride for i in range(num_frames_needed)]
    return frame_indices
```
