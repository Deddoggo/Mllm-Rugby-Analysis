```python
import sys, torch, cv2, transformers, ultralytics, subprocess

print(f"🐍 Python: {sys.version.split()[0]} | 📦 PyTorch: {torch.__version__}")

# Check Apple Silicon GPU (MPS)
mps_ok = torch.backends.mps.is_available()
print(f"🚀 GPU (MPS) Ready: {'✅ YES' if mps_ok else '❌ NO (Using CPU)'}")
if mps_ok:
    print(f"   ↳ Test data transfer to M4 Max GPU: {torch.ones(1).to('mps').device}")

# Check AI & Vision Libraries
print(f"👁️ OpenCV: {cv2.__version__} | 🤖 Transformers: {transformers.__version__} | 🎯 YOLO: {ultralytics.__version__}")

# Check FFmpeg
try:
    ff = subprocess.run(['ffmpeg', '-version'], capture_output=True, text=True).stdout.split()[2]
    print(f"🎬 FFmpeg: ✅ OK (v{ff})")
except Exception:
    print("🎬 FFmpeg: ❌ NOT INSTALLED OR PATH ERROR")
```
